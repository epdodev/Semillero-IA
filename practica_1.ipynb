```python
from huggingface_hub import login
from transformers import pipeline

token = "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"
login(token)

modelo = "google/gemma-2b-it"

modelo_chistes = pipeline(
    "text-generation",
    model=modelo,
    device_map="auto",
    torch_dtype="float16"
)
```

```python
# Paso 1: Definir el prompt
prompt = "Eres un comediante. Escribe un chiste corto con un final inesperado sobre un programador y una IA."

# Paso 2: Generar el chiste controlando la creatividad
resultado = modelo_chistes(
    prompt,
    max_new_tokens=80,    # Longitud máxima de la respuesta
    do_sample=True,       # Habilita el muestreo probabilístico
    temperature=1.2,      # Controla la creatividad (0.1 = conservador, 1.0 = creativo)
    top_p=0.95            # Controla la diversidad de vocabulario
)[0]["generated_text"]

# Paso 3: Mostrar el resultado
print("🎭 Prompt:")
print(prompt)
print("\n🤣 Chiste generado:")
print(resultado)
```
